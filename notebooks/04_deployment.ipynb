```python
import joblib

pipe = joblib.load(
    "../models/sentiment_pipeline.pkl"
)
```

```python
test_reviews = [
    "I absolutely love this product!",
    "This product stopped working after two days.",
    "It is okay, nothing special."
]

predictions = pipe.predict(test_reviews)

for review, prediction in zip(test_reviews, predictions):
    print(f"{prediction}: {review}")
```

```python
review = "That doesn't work at all. Like my old Roborock, a vacuum and mopping robot must be able to be easily integrated into almost every household.And unfortunately that doesn't work at all with the Q10 x5. Only works on hard floors, but also only when everything is open at the top and nothing stands in the way within a radius of about 1-1.5 meters around the station."

prediction = pipe.predict([review])[0]

probabilities = pipe.predict_proba([review])[0]

print("Prediction:", prediction)

for label, probability in zip(pipe.classes_, probabilities):
    print(f"{label}: {probability:.2%}")
```
